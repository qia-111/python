## 欧几里得算法

```python
def gcd(m, n):
    while n != 0:
        m, n = n, m % n
    return m

# 简单测试
print("测试 gcd(12, 8):", gcd(12, 8))  # 应该输出 4
```

```python
# 使用你的函数评估以下数字对的 GCD
print("gcd(2018, 2019) =", gcd(2018, 2019))
print("gcd(1200, 300) =", gcd(1200, 300))
print("gcd(5040, 60) =", gcd(5040, 60))
```

```python
# 观察负数参数会返回什么
print("gcd(-2018, 2019) =", gcd(-2018, 2019))
print("gcd(2018, -2019) =", gcd(2018, -2019))
print("gcd(-1200, 300) =", gcd(-1200, 300))
```

关于负数参数的行为分析：
在 Python 中，负数的取模运算 % 结果仍然是正数，例如 -2018 % 2019 会得到 1。
因此：

gcd(-2018, 2019) 会返回 1，结果正确。

gcd(2018, -2019) 会返回 -1（负数）。

gcd(-1200, 300) 会返回 300。

这种行为有意义吗？
数学上，最大公约数（GCD）通常定义为正数。虽然欧几里得算法在数学推导上可以适用于负数，但直接返回负数结果的实现不够严谨。
如果希望函数总是返回正的 GCD，可以在函数开头加上 m, n = abs(m), abs(n) 进行处理。

## 近似欧拉数 e

```python
def euler_limit(n):
    return (1 + 1/n) ** n

# 测试
print("euler_limit(100):", euler_limit(100))
print("euler_limit(10000):", euler_limit(10000))
```

```python
import math

def euler_infinite_sum(n):
    total = 0
    # 如果 n=1，我们取公式里 k=0 的那一项；如果 n=4，我们取 k=0,1,2,3 的项
    for k in range(n):
        total += 1 / math.factorial(k)
    return total

# 测试
print("euler_infinite_sum(0):", euler_infinite_sum(0))  # 0
print("euler_infinite_sum(1):", euler_infinite_sum(1))  # 1
print("euler_infinite_sum(4):", euler_infinite_sum(4))  # 约 2.667
```

```python
def euler_approx(epsilon):
    true_e = math.exp(1)
    n = 0
    approx = 0
    
    # 当误差大于等于 epsilon 时，继续增加项数
    while abs(approx - true_e) >= epsilon:
        n += 1
        approx = euler_infinite_sum(n)
        
    return approx

# 测试
print("真实 e:", math.exp(1))
print("euler_approx(0.01):", euler_approx(0.01))
print("euler_approx(0.0001):", euler_approx(0.0001))
```

```python
def print_euler_sum_table(n):
    for k in range(1, n + 1):
        print(f"n={k}: {euler_infinite_sum(k)}")

# 测试，打印前 5 行
print_euler_sum_table(5)
```

## 测试整数属性

```python
def is_power_of_2(n):
    if n <= 0:
        return False
    while n % 2 == 0 and n > 1:
        n = n // 2  # 使用整除 //，避免出现浮点数
    return n == 1

# 测试
print("is_power_of_2(8):", is_power_of_2(8))    # True
print("is_power_of_2(10):", is_power_of_2(10))  # False
print("is_power_of_2(1):", is_power_of_2(1))    # True (2^0)
```

```python
def is_power(b, n):
    if n <= 0 or b <= 0:
        return False
    while n % b == 0 and n > 1:
        n = n // b
    return n == 1

# 测试
print("is_power(3, 9):", is_power(3, 9))     # True  (3^2=9)
print("is_power(3, 10):", is_power(3, 10))   # False
print("is_power(5, 125):", is_power(5, 125)) # True  (5^3=125)
print("is_power(2, 8):", is_power(2, 8))     # True  (等同于 is_power_of_2)
```
