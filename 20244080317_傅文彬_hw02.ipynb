## 字符串的乐趣

```python
def is_palindrome(s):
    # 去除空格，并全部转为小写
    clean_s = s.replace(" ", "").lower()
    # 判断是否和反转后的字符串相等
    return clean_s == clean_s[::-1]

# 测试
print(is_palindrome("level"))                     # True
print(is_palindrome("taco cat"))                  # True
print(is_palindrome("T A C O cat"))               # True
print(is_palindrome("rats live on no evil star")) # True
print(is_palindrome("hello"))                     # False
```

```python
def is_abecedarian(s):
    # 去除空格并转小写
    clean_s = s.replace(" ", "").lower()
    # 逐个比较当前字母是否小于等于下一个字母
    for i in range(len(clean_s) - 1):
        if clean_s[i] > clean_s[i+1]:
            return False
    return True

# 测试
print(is_abecedarian("adder"))         # True
print(is_abecedarian("beet"))          # True
print(is_abecedarian("dog"))           # False
print(is_abecedarian("cat"))           # False
print(is_abecedarian("abcd efgh xyz")) # True
```

```python
def remove_vowels(s):
    vowels = "aeiouAEIOU"  # 处理大小写更稳妥
    # 使用列表推导式，保留非元音字母
    result = [char for char in s if char not in vowels]
    return "".join(result)

# 测试
print(remove_vowels("cat"))    # ct
print(remove_vowels("audio"))  # d
print(remove_vowels("aeiou"))  # 空字符串
```

## 列表的乐趣

```python
def list_reverse(lst):
    # 检查输入类型
    if not isinstance(lst, list):
        raise TypeError("输入必须是一个列表 (list)")
    return lst[::-1]  # 使用切片反转

# 测试
print("反转 [1, 2, 3]:", list_reverse([1, 2, 3]))
try:
    list_reverse((1, 2, 3))  # 传入元组会触发错误
except TypeError as e:
    print("捕获异常:", e)
```

```python
def is_sorted(seq):
    # 遍历到倒数第二个元素
    for i in range(len(seq) - 1):
        if seq[i] > seq[i+1]:  # 如果前一个比后一个大，说明不是升序
            return False
    return True

# 测试
print("is_sorted([1, 2, 3]):", is_sorted([1, 2, 3]))       # True
print("is_sorted([1, 3, 2]):", is_sorted([1, 3, 2]))       # False
print("is_sorted('aabbcc'):", is_sorted('aabbcc'))         # True (字符串也是序列)
```

## 作为向量的元组

```python
def vec_scalar_mult(t, s):
    # 检查 t 是否为元组
    if not isinstance(t, tuple):
        raise TypeError("第一个参数必须是元组")
    # 检查 s 是否为整数或浮点数
    if not isinstance(s, (int, float)):
        raise TypeError("第二个参数必须是整数或浮点数")
    # 检查 t 中的元素是否都是数字
    for item in t:
        if not isinstance(item, (int, float)):
            raise TypeError("元组元素必须是整数或浮点数")
    
    # 返回相乘后的新元组
    return tuple(x * s for x in t)

# 测试
print("vec_scalar_mult((1, 2, 3), 2):", vec_scalar_mult((1, 2, 3), 2))
try:
    vec_scalar_mult([1, 2, 3], 2)
except TypeError as e:
    print("捕获异常:", e)
```

```python
def vec_inner_product(t1, t2):
    if not isinstance(t1, tuple) or not isinstance(t2, tuple):
        raise TypeError("两个参数都必须是元组")
    if len(t1) != len(t2):
        raise ValueError("两个向量的长度必须一致")
    
    # 使用 zip 将对应位置的元素相乘后求和
    return float(sum(x * y for x, y in zip(t1, t2)))

# 测试
print("内积 (1,2,3) 和 (4,5,6):", vec_inner_product((1, 2, 3), (4, 5, 6)))
try:
    vec_inner_product((1, 2), (1, 2, 3))
except ValueError as e:
    print("捕获异常:", e)
```

```python
def check_valid_mx(mx):
    # 1. 外层必须是元组
    if not isinstance(mx, tuple):
        return False
    # 2. 每个元素本身也是元组
    for row in mx:
        if not isinstance(row, tuple):
            return False
    
    # 3. 所有内层元组长度相同
    if len(mx) > 0:
        row_len = len(mx[0])
        for row in mx:
            if len(row) != row_len:
                return False
            
            # 4. 每个元素都是整数或浮点数
            for item in row:
                if not isinstance(item, (int, float)):
                    return False
    return True

# 测试
valid_mx = ((1, 2), (3, 4), (5, 6))
invalid_mx = ((1, 2), (3, 4, 5))
print("valid_mx 是有效矩阵吗?", check_valid_mx(valid_mx))     # True
print("invalid_mx 是有效矩阵吗?", check_valid_mx(invalid_mx)) # False
```

```python
def mx_vec_mult(mx, vec):
    # 检查矩阵是否有效
    if not check_valid_mx(mx):
        raise TypeError("无效的矩阵格式")
    # 检查向量是否有效
    if not isinstance(vec, tuple):
        raise TypeError("向量必须是元组")
    
    # 检查维度是否匹配（矩阵的列数 == 向量的长度）
    if len(mx) > 0 and len(mx[0]) != len(vec):
        raise ValueError("矩阵的列数与向量的长度不匹配")
    
    result = []
    for row in mx:
        # 每行与向量求内积
        row_dot = sum(row[i] * vec[i] for i in range(len(vec)))
        result.append(row_dot)
        
    return tuple(result)

# 测试
mx = ((1, 2), (3, 4))
vec = (5, 6)
print("矩阵与向量相乘:", mx_vec_mult(mx, vec))  # ((1*5+2*6), (3*5+4*6)) = (17, 39)
```

## 更多向量的乐趣

```python
class Vector:
    def __init__(self, dim, elements=None):
        # 检查维度类型和是否为非负
        if not isinstance(dim, int):
            raise TypeError("维度必须是整数")
        if dim < 0:
            raise ValueError("维度不能为负数")
        
        self.dim = dim
        
        # 如果未提供元素，默认全 0
        if elements is None:
            self.elements = tuple([0] * dim)
        else:
            if not isinstance(elements, (list, tuple)):
                raise TypeError("元素必须是列表或元组")
            if len(elements) != dim:
                raise ValueError("元素的个数必须与维度一致")
            self.elements = tuple(elements)

    # 为了方便打印查看
    def __repr__(self):
        return f"Vector({self.dim}, {self.elements})"

# 测试
v1 = Vector(3, [1, 2, 3])
v_default = Vector(2) # 只给维度
print("v1:", v1)
print("v_default:", v_default)
```

```python
# 为了测试方便，我们把这些魔法方法直接赋值给 Vector 类
def check_dim(self, other):
    if not isinstance(other, Vector):
        raise TypeError("比较对象必须是 Vector")
    if self.dim != other.dim:
        raise ValueError("维度不同，无法比较")

def eq(self, other):
    check_dim(self, other)
    return self.elements == other.elements

def lt(self, other):
    check_dim(self, other)
    return self.elements < other.elements

def le(self, other):
    check_dim(self, other)
    return self.elements <= other.elements

def gt(self, other):
    check_dim(self, other)
    return self.elements > other.elements

def ge(self, other):
    check_dim(self, other)
    return self.elements >= other.elements

# 绑定到类
Vector.__eq__ = eq
Vector.__lt__ = lt
Vector.__le__ = le
Vector.__gt__ = gt
Vector.__ge__ = ge

# 测试
print("Vector(2, (2,4)) < Vector(2, (2,5)):", Vector(2, (2,4)) < Vector(2, (2,5))) # True
print("Vector(2, (2,4)) == Vector(2, (2,4)):", Vector(2, (2,4)) == Vector(2, (2,4))) # True
try:
    Vector(2, (2,4)) == Vector(3, (2,4,6))
except ValueError as e:
    print("捕获异常:", e)
```

```python
def dot(self, other):
    if not isinstance(other, Vector):
        raise TypeError("参数必须是 Vector 对象")
    if self.dim != other.dim:
        raise ValueError("维度不一致，无法计算内积")
    # 对应元素相乘后求和
    return float(sum(a * b for a, b in zip(self.elements, other.elements)))

Vector.dot = dot

# 测试
v_a = Vector(2, (1, 2))
v_b = Vector(2, (3, 4))
print("v_a.dot(v_b):", v_a.dot(v_b)) # 1*3 + 2*4 = 11
```

```python
def mul(self, other):
    # 如果是数字（标量）
    if isinstance(other, (int, float)):
        return Vector(self.dim, tuple(x * other for x in self.elements))
    # 如果是向量
    elif isinstance(other, Vector):
        if self.dim != other.dim:
            raise ValueError("维度不一致，无法相乘")
        return Vector(self.dim, tuple(a * b for a, b in zip(self.elements, other.elements)))
    else:
        raise TypeError("不支持的类型")

def rmul(self, other):
    # 处理 2 * v 的情况，直接调用绑定好的 __mul__
    return self.__mul__(other)

Vector.__mul__ = mul
Vector.__rmul__ = rmul

# 测试
v = Vector(2, (2, 3))
w = Vector(2, (4, 5))

print("2 * v:", 2 * v)         # Vector(2, (4, 6))
print("v * 2:", v * 2)         # Vector(2, (4, 6))
print("v * w:", v * w)         # Vector(2, (8, 15))
```
