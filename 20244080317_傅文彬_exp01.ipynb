## 实验2：函数、条件、循环与递归

```python
def score_level(score):
    if not (0 <= score <= 100):
        raise ValueError("成绩必须在 0 到 100 之间")
    if score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"

# 测试常规成绩
print(score_level(95))   # 优秀
print(score_level(85))   # 良好
print(score_level(75))   # 及格
print(score_level(50))   # 不及格

# 测试异常
try:
    score_level(105)
except ValueError as e:
    print("捕获异常:", e)
```

```python
# for 循环
total_for = 0
for i in range(1, 101):
    total_for += i

# while 循环
total_while = 0
i = 1
while i <= 100:
    total_while += i
    i += 1

print("for 循环结果:", total_for)
print("while 循环结果:", total_while)
print("两者相等吗:", total_for == total_while)
```

```python
# 递归版
def factorial_recursive(n):
    if n == 0:
        return 1
    return n * factorial_recursive(n - 1)

# 迭代版
def factorial_iterative(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

# 比较
for n in [0, 1, 5, 10]:
    print(f"n={n}, 递归={factorial_recursive(n)}, 迭代={factorial_iterative(n)}")
```

```python
def summary(values):
    min_val = min(values)
    max_val = max(values)
    avg_val = sum(values) / len(values)
    return (min_val, max_val, avg_val)

# 测试与解包
data = [85, 58, 92, 76, 60, 89]
min_v, max_v, avg_v = summary(data)
print("最小值:", min_v)
print("最大值:", max_v)
print("平均值:", round(avg_v, 2))
```

```python
def sum_all(*args):
    return sum(args)

print("无参数:", sum_all())             # 0
print("单个参数:", sum_all(5))          # 5
print("多个参数:", sum_all(1, 2, 3, 4)) # 10
```

## 实验3：列表、字典、元组与基础统计

```python
names = ["Alice", "Bob", "Carol", "David", "Eva", "Frank"]
scores = [85, 58, 92, 76, 60, 89]

# 使用 zip() 构建姓名到成绩的字典
score_dict = dict(zip(names, scores))
print("字典:", score_dict)

# 使用 items() 逐项输出姓名和成绩
for name, score in score_dict.items():
    print(name, score)
```

```python
# 使用 enumerate() 输出序号（从1开始）、姓名和成绩
for index, (name, score) in enumerate(score_dict.items(), start=1):
    print(f"序号: {index}, 姓名: {name}, 成绩: {score}")
```

```python
# 提取及格学生的姓名
pass_names = [name for name, score in score_dict.items() if score >= 60]
print("及格学生:", pass_names)

# 生成所有成绩的平方列表
score_squares = [score ** 2 for score in scores]
print("成绩平方:", score_squares)
```

```python
# 计算统计指标
count = len(scores)
avg_score = sum(scores) / count
max_score = max(scores)
min_score = min(scores)
pass_rate = len(pass_names) / count

# 格式化输出（平均分两位小数，及格率百分数）
print(f"人数: {count}")
print(f"平均分: {avg_score:.2f}")
print(f"最高分: {max_score}")
print(f"最低分: {min_score}")
print(f"及格率: {pass_rate:.1%}")
```

## 实验4：文件读写、异常处理与模块

```python
import os

# 1. 使用 os.getcwd() 查看工作目录
print("当前工作目录:", os.getcwd())

# 2. 使用 os.path.join() 构造相对路径
data_dir = "exp01_data"
file_path = os.path.join(data_dir, "students.txt")

# 3. 使用 os.makedirs(..., exist_ok=True) 创建数据目录
os.makedirs(data_dir, exist_ok=True)
print("数据目录已创建:", data_dir)
```

```python
# 写入数据 (UTF-8)
data_content = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete"""

with open(file_path, "w", encoding="utf-8") as f:
    f.write(data_content)

# 读取文件并跳过表头
with open(file_path, "r", encoding="utf-8") as f:
    lines = f.readlines()
    for line in lines[1:]:  # 跳过第一行表头
        print(line.strip())
```

```python
def parse_student(line):
    parts = line.strip().split(",")
    
    # 检查字段数量
    if len(parts) != 2:
        raise ValueError("字段数量不足，应为 name,score")
    
    name, score_str = parts[0].strip(), parts[1].strip()
    
    # 检查姓名是否为空
    if not name:
        raise ValueError("姓名不能为空")
    
    # 检查成绩是否为整数
    try:
        score = int(score_str)
    except ValueError:
        raise ValueError(f"成绩必须是整数，当前为 '{score_str}'")
    
    # 检查成绩范围
    if not (0 <= score <= 100):
        raise ValueError(f"成绩必须在 0~100 范围内，当前为 {score}")
    
    return {"name": name, "score": score}
```

```python
valid_records = []
invalid_records = []

with open(file_path, "r", encoding="utf-8") as f:
    header = f.readline()  # 跳过表头
    for line_num, line in enumerate(f, start=2):  # 从第2行开始
        if not line.strip():
            continue
        try:
            record = parse_student(line)
            valid_records.append(record)
        except ValueError as e:
            invalid_records.append({"line": line_num, "error": str(e)})
            print(f"第 {line_num} 行无效: {e}")

print("\n有效记录:", valid_records)
print("无效记录:", invalid_records)
```

```python
# 计算统计数据
valid_count = len(valid_records)
invalid_count = len(invalid_records)

if valid_count > 0:
    avg_score = sum(r["score"] for r in valid_records) / valid_count
    pass_rate = sum(1 for r in valid_records if r["score"] >= 60) / valid_count
else:
    avg_score = 0
    pass_rate = 0

report_content = f"""有效记录数: {valid_count}
无效记录数: {invalid_count}
平均分: {avg_score:.2f}
及格率: {pass_rate:.1%}
"""

report_path = os.path.join(data_dir, "report.txt")

# 演示 w 模式（覆盖原有内容）
with open(report_path, "w", encoding="utf-8") as f:
    f.write("=== 报告 (w模式) ===\n")
    f.write(report_content)

# 演示 a 模式（在末尾追加）
with open(report_path, "a", encoding="utf-8") as f:
    f.write("\n=== 追加内容 (a模式) ===\n")
    f.write("这行是使用 a 模式追加的，原来的内容不会被覆盖。\n")

print("w 模式会覆盖原有内容，a 模式会在末尾追加内容。")
print(f"报告已写入: {report_path}")
```

## 实验5：类、继承与对象复制

```python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        return f"学生: {self.name}, 成绩: {self.score}, 及格状态: {'及格' if self.is_passed() else '不及格'}"
```

```python
# 为实验4中的每条有效记录创建一个 Student 对象
student_objects = [Student(r["name"], r["score"]) for r in valid_records]

# 输出对象信息和及格状态
for obj in student_objects:
    print(obj)
```

```python
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        # 通过 super().__init__() 初始化父类属性
        super().__init__(name, score)
        self.research_topic = research_topic

    # 重写 __str__()
    def __str__(self):
        return f"研究生: {self.name}, 成绩: {self.score}, 研究方向: {self.research_topic}, 及格状态: {'及格' if self.is_passed() else '不及格'}"

# 测试
grad_student = GraduateStudent("Alice", 85, "人工智能")
print(grad_student)
```

```python
import copy

# 每次比较都从新建的原始对象开始
original = {"name": "Alice", "scores": [85, 90]}

# 1. 直接赋值
print("--- 直接赋值 ---")
obj1 = original
obj1["scores"].append(95)  # 修改内部列表
print("原对象:", original)
print("直接赋值对象:", obj1)
# 结论：直接赋值只复制了引用，修改内部列表会影响原对象。

# 重新新建原始对象进行下一轮比较
original = {"name": "Alice", "scores": [85, 90]}

# 2. 浅拷贝 copy.copy()
print("\n--- 浅拷贝 copy.copy() ---")
obj2 = copy.copy(original)
obj2["scores"].append(95)  # 修改内部列表
print("原对象:", original)
print("浅拷贝对象:", obj2)
# 结论：浅拷贝只复制最外层，内部嵌套的列表还是共享的，修改内部列表会影响原对象。

# 重新新建原始对象进行下一轮比较
original = {"name": "Alice", "scores": [85, 90]}

# 3. 深拷贝 copy.deepcopy()
print("\n--- 深拷贝 copy.deepcopy() ---")
obj3 = copy.deepcopy(original)
obj3["scores"].append(95)  # 修改内部列表
print("原对象:", original)
print("深拷贝对象:", obj3)
# 结论：深拷贝会递归复制所有层级，修改嵌套列表不会影响原对象。
```
